#Задача 7. Средняя цена по городам
Посчитайте среднюю цену квартиры для каждого города.

Сделайте вывод:

В каком городе самая высокая средняя цена?
В каком городе самая низкая средняя цена?
Насколько сильно отличаются цены между городами?

```python
avg_price_new=df.groupby("city")["price_rub"].mean().sort_values(ascending=False).reset_index().rename(columns={"price_rub": "avg_price"})
avg_price_new
```

```python
size_price = avg_price_new["avg_price"].max() / avg_price_new["avg_price"].min()
print(size_price)
```

#Ответ
Самая высокая средняя цена в г.Санкт-Петербург. Самая низкая средняя цена в г.Самара.
Между городами цены отличаются почти в 4 раза.
