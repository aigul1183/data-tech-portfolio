```python
import requests
from bs4 import BeautifulSoup

url = "https://skillfactory.ru/"

response = requests.get(url, timeout=10)

print("Код ответа:", response.status_code)

soup = BeautifulSoup(response.text, "html.parser")

title = soup.find("title")
h1 = soup.find("h1")
paragraph = soup.find("p")
link = soup.find("a")

print("title:", title.get_text(strip=True) if title else "Не найдено")
print("h1:", h1.get_text(strip=True) if h1 else "Не найдено")
print("p:", paragraph.get_text(strip=True) if paragraph else "Не найдено")

if link:
    print("Текст ссылки:", link.get_text(strip=True))
    print("Адрес ссылки:", link.get("href"))
else:
    print("Ссылка не найдена")
```

В результате парсинга выведены такие данные, как:

* название страницы
* заголовок 1-го уровня
* первый параграф
* ссылка на страницу




```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

url = "https://skillfactory.ru/"

headers = {
    "User-Agent": "Mozilla/5.0"
}

try:
    response = requests.get(url, headers=headers, timeout=10)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = soup.find_all("a")

    count = 0
    for link in links:
        if count >= 7:
            break

        text = link.get_text(strip=True)
        href = link.get("href")

        if href:
            full_url = urljoin(url, href)
            print("Текст:", text if text else "[без текста]")
            print("Ссылка:", full_url)
            print("-" * 40)
            count += 1

except requests.exceptions.RequestException as error:
    print("Ошибка запроса:", error)
```

В результате выведены все ссылки, которые есть по заданному url.
