#Рейтинг ESRB

1. Необходимо вывести категории рейтинга.

```python
rating_summary = df['rating'].value_counts(dropna=False).reset_index()
rating_summary.columns = ['rating', 'count']
print(rating_summary)
```

2. В столбце "rating" 6766 пропусков.
Необходимо определить пропуски среди игр по рейтингу в разрезе регионов: Северной Америки, Европы и Японии.


```python
# Пропуски рейтинга среди игр, продававшихся в Северной Америке
na_mask = df['na_sales'].notna() & (df['na_sales'] > 0)
missing_rating_in_na = df.loc[na_mask, 'rating'].isna().sum()
percent_na = (missing_rating_in_na / na_mask.sum() * 100) if na_mask.sum() > 0 else 0

# Пропуски рейтинга среди игр, продававшихся в Европе
eu_mask = df['eu_sales'].notna() & (df['eu_sales'] > 0)
missing_rating_in_eu = df.loc[eu_mask, 'rating'].isna().sum()
percent_eu = (missing_rating_in_eu / eu_mask.sum() * 100) if eu_mask.sum() > 0 else 0

# Пропуски рейтинга среди игр, продававшихся в Японии
jp_mask = df['jp_sales'].notna() & (df['jp_sales'] > 0)
missing_rating_in_jp = df.loc[jp_mask, 'rating'].isna().sum()
percent_jp = (missing_rating_in_jp / jp_mask.sum() * 100) if jp_mask.sum() > 0 else 0

print("Пропусков rating среди игр с продажами:")
print(f"NA: {missing_rating_in_na} ({percent_na:.2f} %)")
print(f"EU: {missing_rating_in_eu} ({percent_eu:.2f} %)")
print(f"JP: {missing_rating_in_jp} ({percent_jp:.2f} %)")
```

Доля пропусков в рейтинге среди игр: в Северной Америке - 24,9%, в Европе - 28%, в Японии - 64,5%.

```python
leader_na = df.groupby('rating').agg(na_total=('na_sales', 'sum')).reset_index()
leader_na.sort_values('na_total', ascending=False).head(2)
```

```python
leader_eu = df.groupby('rating').agg(eu_total=('eu_sales', 'sum')).reset_index()
leader_eu.sort_values('eu_total', ascending=False).head(2)
```

```python
leader_jp = df.groupby('rating').agg(jp_total=('jp_sales', 'sum')).reset_index()
leader_jp.sort_values('jp_total', ascending=False).head(2)
```
