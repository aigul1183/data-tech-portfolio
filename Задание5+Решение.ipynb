##Задача 2. Работа с датой регистрации
Преобразуйте столбец registration_date к типу даты.

Добавьте новые столбцы:

*   registration_weekday — день недели регистрации;
*   registration_month — месяц регистрации.

Постройте гистограмму количества регистраций по месяцам.

````python
import pandas as pd

data['registration_date'] = pd.to_datetime(data['registration_date'])
````

````python
data['registration_weekday'] = data['registration_date'].dt.day_name()
````

````python
data['registration_month'] = data['registration_date'].dt.month_name()
````

````python
data[['registration_date', 'registration_weekday', 'registration_month']].head()
````

````python
data.info()
````

````python
data['registration_month'].unique()
````

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.histplot(data=data, x='registration_month', discrete=True, color='steelblue')
#sns.histplot(data=data, x='registration_month',  color='steelblue')  # без discrete=True
plt.title('Количество регистраций по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Число регистраций')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

##Вывод:
Колонка "registration_date" преобразована к типу даты: datetime64[ns]. Новые столбцы "registration_weekday" и "registration_month" к основной таблице добавлены. В колонке "registration_month" представлены месяцы с января по июнь. На гистограмме показано количество регистраций за данный период.

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.histplot(data=data, x='course_price', bins=20, color='steelblue', edgecolor='black')

plt.title('Распределение исходной цены курса')
plt.xlabel('Course price')
plt.ylabel('Количество записей (студентов)')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.histplot(data=data, x='final_price', bins=20, color='blue', edgecolor='red')

plt.title('Распределение итоговой цены курса (final_price)')
plt.xlabel('Итоговая цена')
plt.ylabel('Количество записей (студентов)')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
plt.figure(figsize=(8, 4))
sns.histplot(data=data, x='course_price', bins=20, color='lightblue', label='Course price', alpha=0.7)
sns.histplot(data=data, x='final_price', bins=20, color='darkblue', label='Final price', alpha=0.7)

plt.title('Course price vs Final price')
plt.xlabel('Цена')
plt.ylabel('Количество записей')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
mean_price = data['final_price'].mean()
median_price = data['final_price'].median()

print(f'Среднее значение final_price: {mean_price:.2f}')
print(f'Медиана final_price: {median_price}')
````

##Вывод

Среднее и медиана почти совпадают, что означает симметричное распределение цен: нет большого числа очень дорогих покупок, которые искажали бы среднее значение и в распределении "final_price" нет сильных перекосов за счет скидок.




## Задача 4. Анализ прохождения обучения

Создайте новый столбец:

```text
completion_rate
```

Он должен показывать долю пройденных уроков:

```text
lessons_completed / lessons_total
```

После этого:

1. Постройте гистограмму `completion_rate`.
2. Найдите среднее и медиану `completion_rate`.
3. Сравните средний `completion_rate` для тех, кто получил сертификат, и тех, кто не получил.


````python
data['completion_rate'] = (data['lessons_completed'] / data['lessons_total'])*100
data['completion_rate']
````

````python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.histplot(data=data, x='completion_rate', bins=20, color='teal', edgecolor='black')

plt.title('Распределение процента завершённых уроков (completion_rate)')
plt.xlabel('Completion rate (%) — доля пройденных уроков')
plt.ylabel('Количество студентов')
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
mean_com_rate = data['completion_rate'].mean()
median_com_rate = data['completion_rate'].median()

print(f'Среднее completion_rate: {mean_com_rate:.2f}%')
print(f'Медиана completion_rate: {median_com_rate:.2f}%')
````

````python

mean_by_cert = data.groupby('certificate_received')['completion_rate'].mean().round(2)
print(mean_by_cert)
````

##Вывод
Новый столбец "completion_rate" создан(%) - доля пройденных уроков.

Среднее значение по данному столбцу 75.24% — это «общий средний уровень», который учитывает всех, включая тех, кто почти ничего не сделал.
Медиана 81.93% — это уровень «типичного» пользователя: половина студентов завершила курс на 81.93% или выше.
Получается, что «типичная» картина лучше общей средней — значит, есть ощутимая доля тех, кто сильно не дотягивает до середины.

False (49.71 %) - это те, кто сертификат не получил, в среднем прошли чуть меньше половины курса.
True (90.40 %) - те, кто получил сертификат, дойдя почти до конца курса.

## Задача 5. Поиск закономерностей в продажах

С помощью сводных таблиц ответьте на вопросы:

1. Какая категория курсов приносит больше всего выручки по `final_price`?
2. У какого источника привлечения (`lead_source`) самая высокая средняя итоговая цена покупки?
3. У какого курса самая высокая средняя удовлетворённость `satisfaction_score`?
4. Постройте график зависимости медианной итоговой цены от месяца регистрации.
5. Сделайте 2–3 вывода по результатам анализа.

````python
course_total = data.groupby('course_category').agg(final_price_total=('final_price', 'sum'), final_price_count=('final_price', 'count') ).reset_index()
course_total.sort_values('final_price_total', ascending=False)
````

##Вывод
Категория "Программирование" приносит больше всего выручки, данный факт подтверждает сумма и количество купленных курсов.

````python
data['final_price'].unique()
````

````python
lead_source_stats = data.groupby('lead_source').agg(final_price_mean=('final_price', 'mean'), final_price_count=('final_price', 'count'),    final_price_total=('final_price', 'sum') ).reset_index()
lead_source_stats['final_price_mean'] = lead_source_stats['final_price_mean'].round(0)

lead_source_stats.sort_values('final_price_mean', ascending=False)
````

##Вывод

Канал "Рекомендация" даёт самый высокий средний чек, но из‑за небольшого числа сделок выводы по нему не объективны.

Канал "Блог" показывает стабильно высокий чек.

Канал "Поиск" показывает чек чуть ниже, но кол-во сделок выше, чем в двух предыдущих каналах, а также большая суммарная выручка.

Рекламный канал приносит больше всего сделок. Не смотря на то, что средний чек тут ниже, общая выручка по этому каналу - самая большая.

Соцсети дают самый низкий средний чек, но кол-во сделок и суммарная выручка больше при сравнении с каналами "Рекомендация" и "Блог".

````python
course_name_new= data.groupby('course_name').agg(satisfaction_score_mean=('satisfaction_score', 'mean'), satisfaction_score_count=('satisfaction_score', 'count') ).reset_index()
course_name_new.sort_values('satisfaction_score_mean', ascending=False)
````
